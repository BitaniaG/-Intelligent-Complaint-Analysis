```python
from src.retrieval.retriever import ComplaintRetriever
from src.generation.generator import ComplaintGenerator
from src.rag.rag_pipeline import ComplaintRAGPipeline

retriever = ComplaintRetriever(
    index_path="../vector_db/faiss_index",
    metadata_path="../vector_db/faiss_index_meta.pkl"
)
generator = ComplaintGenerator()
rag = ComplaintRAGPipeline(retriever, generator)
```

```python
evaluation_questions = [
    "Why are customers unhappy with credit cards?",
    "What recurring issues do customers report about money transfers?",
    "What problems do customers face with savings accounts?",
    "Are personal loan complaints mostly about interest rates or repayment?",
    "What common themes appear across all financial products?"
]
```

Purpose:

Defines what we will evaluate

These questions satisfy the rubric

```python
results = []

for question in evaluation_questions:
    answer, sources = rag.answer(question, top_k=5)

    results.append({
        "question": question,
        "answer": answer,
        "num_sources": len(sources),
        "products": list(set([s["product"] for s in sources]))
    })
```

Purpose:

Executes the RAG pipeline

Stores outputs for evaluation

Keeps answers + metadata together

```python
for r in results:
    print("=" * 80)
    print("QUESTION:", r["question"])
    print("\nANSWER:\n", r["answer"])
    print("\nSOURCE PRODUCTS:", r["products"])
```
