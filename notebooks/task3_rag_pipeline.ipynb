```python
from src.retrieval.retriever import ComplaintRetriever

retriever = ComplaintRetriever(
    index_path="../vector_db/faiss_index",
    metadata_path="../vector_db/faiss_index_meta.pkl"
)

query = "Why are customers unhappy with credit cards?"

results = retriever.retrieve(query, top_k=5)

for r in results:
    print("Product:", r["product"])
    print("Complaint ID:", r["complaint_id"])
    print("Text:", r["text"][:300])
    print("-" * 80)
```

```python
retriever = ComplaintRetriever(
    index_path="../vector_db/faiss_index",
    metadata_path="../vector_db/faiss_index_meta.pkl"
)

generator = ComplaintGenerator()

rag_pipeline = ComplaintRAGPipeline(
    retriever=retriever,
    generator=generator
)
```

```python
question = "What are the most common customer complaints about credit cards?"

answer, sources = rag_pipeline.answer(
    question=question,
    top_k=5
)

print("ANSWER:")
print(answer)

print("\nSOURCES:")
for s in sources:
    print(f"- Product: {s['product']}, Complaint ID: {s['complaint_id']}")
```
